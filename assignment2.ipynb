```python


import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
uploaded=files.upload()

filename = list(uploaded.keys())[0]
img = cv2.imread(filename)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)



R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
gray_mean = (R.astype(float) + G.astype(float) + B.astype(float)) / 3
gray_mean = gray_mean.astype(np.uint8)



plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original RGB Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(gray_mean, cmap="gray")
plt.title("Grayscale - Mean")
plt.axis("off")
plt.show()

```

```python
# ---------------- CASE 2 ----------------

wr = float(input("Enter weight of R: "))
wg = float(input("Enter weight of G: "))
wb = float(input("Enter weight of B: "))

# Check weights
if wr < 0 or wg < 0 or wb < 0:
    print("Weights cannot be negative.")

elif abs((wr + wg + wb) - 1) > 0.0001:
    print("Sum of weights must be equal to 1.")

else:
    # Weighted grayscale
    gray_weighted = (
        wr * R.astype(float) +
        wg * G.astype(float) +
        wb * B.astype(float)
    )

    gray_weighted = gray_weighted.astype(np.uint8)

    # Display
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    plt.imshow(img)
    plt.title("Original RGB Image")
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.imshow(gray_weighted, cmap="gray")
    plt.title("Grayscale - Weighted")
    plt.axis("off")
    plt.show()
```
